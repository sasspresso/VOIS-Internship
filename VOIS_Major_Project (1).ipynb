```python
outlier_cols = [
    'Yield_Tonnes_Ha',
    'Profit_per_Ha_INR',
    'Water_Efficiency_t_per_1000m3'
]

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    outliers = df[
        (df[col] < Q1 - 1.5 * IQR) |
        (df[col] > Q3 + 1.5 * IQR)
    ]

    print(col, ':', len(outliers), 'outliers')
```

```python
# Q9: Outlier distribution across seasons

for metric in [
    'Yield_Tonnes_Ha',
    'Profit_per_Ha_INR',
    'Water_Efficiency_t_per_1000m3'
]:
    Q1 = df[metric].quantile(0.25)
    Q3 = df[metric].quantile(0.75)
    IQR = Q3 - Q1

    outlier_mask = (
        (df[metric] < Q1 - 1.5 * IQR) |
        (df[metric] > Q3 + 1.5 * IQR)
    )

    print(f"\n{metric}")
    print(df.loc[outlier_mask, 'Season'].value_counts())
```

```python
from scipy.stats import kruskal

metrics = [
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Revenue_per_Ha_INR',
    'Profit_per_Ha_INR',
    'Water_Efficiency_t_per_1000m3'
]

for metric in metrics:
    groups = [
        df.loc[df['Season'] == season, metric].dropna()
        for season in ['Kharif', 'Rabi', 'Zaid']
    ]

    stat, p = kruskal(*groups)

    print(f"{metric}: H = {stat:.3f}, p = {p:.4f}")
```

```python
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests

seasons = ['Kharif', 'Rabi', 'Zaid']
results = []

for metric in metrics:
    for i in range(len(seasons)):
        for j in range(i + 1, len(seasons)):
            a = df.loc[df['Season'] == seasons[i], metric].dropna()
            b = df.loc[df['Season'] == seasons[j], metric].dropna()

            u, p = mannwhitneyu(a, b, alternative='two-sided')
            results.append([metric, seasons[i], seasons[j], u, p])

posthoc = pd.DataFrame(
    results,
    columns=['Metric', 'Season_1', 'Season_2', 'U', 'p_value']
)

posthoc['p_adjusted'] = multipletests(
    posthoc['p_value'],
    method='bonferroni'
)[1]

posthoc['Significant'] = posthoc['p_adjusted'] < 0.05

posthoc.round(4)
```

```python
from scipy.stats import chi2_contingency

for variable in ['Crop', 'Irrigation_Method']:
    table = pd.crosstab(df['Season'], df[variable])
    chi2, p, dof, expected = chi2_contingency(table)

    print(f"\n{variable}")
    print(f"Chi-square = {chi2:.3f}")
    print(f"p-value = {p:.4f}")
```

```python
from scipy.stats import spearmanr

env_vars = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_pH',
    'Soil_Moisture_pct'
]

outcomes = [
    'Yield_Tonnes_Ha',
    'Profit_per_Ha_INR',
    'Water_Efficiency_t_per_1000m3'
]

correlation_results = []

for env in env_vars:
    for outcome in outcomes:
        temp = df[[env, outcome]].dropna()
        rho, p = spearmanr(temp[env], temp[outcome])

        correlation_results.append([
            env, outcome, rho, p
        ])

spearman_results = pd.DataFrame(
    correlation_results,
    columns=['Environmental_Variable', 'Outcome', 'Spearman_Rho', 'p_value']
)

spearman_results.round(4)
```

```python
season_corr = []

for season in ['Kharif', 'Rabi', 'Zaid']:
    subset = df[df['Season'] == season]

    for env in env_vars:
        for outcome in outcomes:
            temp = subset[[env, outcome]].dropna()
            rho, p = spearmanr(temp[env], temp[outcome])

            season_corr.append([
                season, env, outcome, rho, p
            ])

season_corr_results = pd.DataFrame(
    season_corr,
    columns=['Season', 'Environmental_Variable',
             'Outcome', 'Spearman_Rho', 'p_value']
)

season_corr_results.round(4)
```

## 10. Key Insights from Observed Seasonal Differences

- **Kharif** shows the strongest overall performance, with the highest yield, production, revenue/ha, profit/ha and water efficiency.
- **Zaid** shows the weakest overall performance, particularly in yield and profitability.
- Seasonal differences are **statistically significant** across the major performance indicators.
- The **Kharif > Rabi > Zaid** pattern is broadly consistent across crops and states.
- Environmental relationships vary by season, notably the **reversal in the rainfall–yield association** across seasons.
- **Crop and irrigation composition are not significantly associated with season**, suggesting that seasonal differences are not simply explained by changes in these categories.

## 11. Conclusions, Recommendations & Limitations

### Conclusion
- **Kharif** shows the strongest overall agricultural and economic performance, while **Zaid** shows the weakest.
- Seasonal differences are statistically significant across the major performance indicators.
- The **Kharif > Rabi > Zaid** pattern is broadly consistent across crops and states.
- Environmental–performance relationships vary across seasons, particularly the rainfall–yield association.

### Recommendations
- Adopt **season-specific production and resource planning**.
- Investigate factors contributing to weaker **Zaid profitability and performance**.
- Consider seasonal environmental conditions when planning irrigation and farm inputs.
- Use crop- and region-level analysis to target seasonal interventions.

### Limitations
- The observational dataset supports **associations, not causal conclusions**.
- Mathematically related variables should not be treated as independent predictors.
- State/District combinations may not represent real-world geography; geographic findings should therefore be interpreted as **dataset-level patterns**.
- IQR outliers indicate unusual observations but do not necessarily represent data errors.
